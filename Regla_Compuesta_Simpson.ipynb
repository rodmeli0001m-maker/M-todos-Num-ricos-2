# Regla compuesta de Simpson

## Introducción

La regla compuesta de Simpson es un método de integración numérica que permite aproximar el valor de una integral definida.

El método divide el intervalo $[a,b]$ en $n$ subintervalos de igual longitud, donde $n$ debe ser un número par.

El tamaño de cada subintervalo se calcula mediante:

$$
h=\frac{b-a}{n}
$$

## Fórmula de Simpson compuesta

La regla compuesta de Simpson aproxima la integral definida mediante la siguiente expresión:

$$
\int_a^b f(x)\,dx
\approx
\frac{h}{3}
\left[
f(a)+f(b)
+4\sum_{\substack{i=1\\i\ impar}}^{n-1}f(x_i)
+2\sum_{\substack{i=2\\i\ par}}^{n-2}f(x_i)
\right]
$$

donde:

- $a$ y $b$ son los extremos del intervalo.
- $n$ es el número de subintervalos y debe ser par.
- $h$ es el tamaño de cada subintervalo.
- $x_i=a+ih$ son los puntos donde se evalúa la función.
- Los valores correspondientes a índices impares se multiplican por $4$.
- Los valores correspondientes a índices pares se multiplican por $2$.

Algoritmo

Para aplicar la regla compuesta de Simpson se siguen los siguientes pasos:

1. Introducir la función f(x) y los extremos del intervalo a y b.

2. Introducir el número de subintervalos n, el cual debe ser positivo y par.

3. Calcular el tamaño de cada subintervalo:
   
   h = (b - a) / n

4. Calcular los valores de la función en los extremos a y b.

5. Recorrer los puntos interiores del intervalo.

6. Sumar por separado los valores correspondientes a índices impares y pares.

7. Multiplicar la suma de los valores impares por 4 y la suma de los valores pares por 2.

8. Aplicar la fórmula de Simpson compuesta.

9. Mostrar el resultado aproximado de la integral.

```python
from math import *

def simpson_compuesta():

    try:

        # Entrada de datos
        funcion = input("Escriba f(x) = ")
        a = float(input("Ingrese el extremo a: "))
        b = float(input("Ingrese el extremo b: "))
        n = int(input("Ingrese n (entero positivo y par): "))


        if a >= b:
            raise ValueError("Debe cumplirse que a < b.")

        if n <= 0 or n % 2 != 0:
            raise ValueError("n debe ser un entero positivo y par.")

        # Definición de la función
        f = lambda x: eval(funcion)

        # Cálculo del tamaño de los subintervalos
        h = (b - a) / n

        # Se obtiene la suma de los valores en los extremos
        XI0 = f(a) + f(b)

        # Se inicializan las sumas
        XI1 = 0
        XI2 = 0

        # Se recorren los puntos interiores del intervalo
        for i in range(1, n):

            # Cálculo del punto correspondiente
            x = a + i * h

            # Se separan los términos pares e impares
            if i % 2 == 0:
                XI2 = XI2 + f(x)
            else:
                XI1 = XI1 + f(x)

        # Se aplica la fórmula de Simpson compuesta
        XI = h * (XI0 + 2 * XI2 + 4 * XI1) / 3

        # Se muestra el resultado de la integral
        print("\nResultado:")
        print("XI =", XI)

    except ValueError as e:
        print("Error:", e)

    finally:
        # Indica que el programa terminó su ejecución
        print("\nOperación finalizada.")


simpson_compuesta()
```
