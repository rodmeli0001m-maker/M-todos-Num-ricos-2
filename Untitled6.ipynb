# Cuadratura adaptable

**Introducción**

La cuadratura adaptable es un método de integración numérica que permite aproximar el valor de una integral definida.

Este método utiliza la regla de Simpson y divide el intervalo en partes más pequeñas cuando necesita mejorar la precisión del resultado.

En este trabajo se utilizará Python para aproximar la siguiente integral en el intervalo de 1 a 3, con una tolerancia de 0.0001:

$$
I=\int_1^3 \frac{100}{x^2}\sin\left(\frac{10}{x}\right)dx
$$

# Fórmula de Simpson

La regla de Simpson permite aproximar una integral definida mediante la siguiente expresión:

$$
S(a,b)=\frac{b-a}{6}
\left[f(a)+4f(c)+f(b)\right]
$$

Donde el punto medio se calcula como:

$$
c=\frac{a+b}{2}
$$

Donde:

- a y b: son los límites del intervalo.
- c: es el punto medio del intervalo.
- f(x): es la función que se desea integrar.
- S(a,b): es la aproximación de la integral mediante la regla de Simpson.

Cuadratura adaptable

El método calcula primero la aproximación en el intervalo completo. Después, divide el intervalo en dos partes y calcula una nueva aproximación en cada una.

$$
S_{\text{total}}=S(a,c)+S(c,b)
$$

Posteriormente, compara ambas aproximaciones para estimar el error:

$$
E=\left|S_{\text{total}}-S(a,b)\right|
$$

Si se cumple el criterio de tolerancia, se acepta el resultado. De lo contrario, el intervalo se subdivide nuevamente hasta alcanzar la precisión deseada.

# Algoritmo

Para aproximar la integral mediante cuadratura adaptable, se siguen los siguientes pasos:

1. Definir la función (f(x)), los límites de integración y la tolerancia.
2. Calcular la aproximación inicial utilizando la regla de Simpson.
3. Encontrar el punto medio del intervalo.
4. Calcular las aproximaciones de Simpson en las dos mitades.
5. Comparar la suma de las aproximaciones con el resultado inicial.
6. Verificar si se cumple el criterio de tolerancia.
7. Si no se cumple, dividir nuevamente el intervalo y repetir el procedimiento.
8. Sumar las aproximaciones obtenidas y mostrar el resultado final.

A continuacón se muestra el programa para la resolución del método de la cuadratura adaptable:

```python
import math

# La cuadratura adaptable es un método numérico para aproximar
# integrales definidas. A diferencia de la regla de Simpson tradicional,
# este método subdivide automáticamente el intervalo en regiones
# donde la función requiere mayor precisión, permitiendo obtener
# resultados más exactos con menos evaluaciones.

# Regla de Simpson
def Simpson(f, a, b):
    c = (a + b) / 2
    return (b - a) * (f(a) + 4 * f(c) + f(b)) / 6

# Cuadratura adaptable
def Adaptive_Simpson(f, a, b, tol):
    s = Simpson(f, a, b)

    # Punto medio
    c = (a + b) / 2

    # Simpson en dos mitades
    s_left = Simpson(f, a, c)
    s_right = Simpson(f, c, b)

    # Estimación del error
    error = abs((s_left + s_right) - s)

    # Verificando tolerancia
    if error < 10 * tol:
        return s_left + s_right

    # Subdividir recursivamente
    izquierda = Adaptive_Simpson(f, a, c, tol / 2)
    derecha = Adaptive_Simpson(f, c, b, tol / 2)

    return izquierda + derecha

# Función principal para cualquier ejercicio
def cuadratura_adaptable():
    try:
        # Entrada de datos
        funcion = input("Escriba f(x) = ")
        a = float(input("Ingrese el extremo a: "))
        b = float(input("Ingrese el extremo b: "))
        tolerancia = float(input("Ingrese la tolerancia: "))

        # Validación de datos
        if a >= b:
            raise ValueError("El extremo a debe ser menor que b.")

        if tolerancia <= 0:
            raise ValueError("La tolerancia debe ser positiva.")

        # Definición de la función ingresada
        f = lambda x: eval(funcion, {"__builtins__": {}}, {
            "x": x,
            "sin": math.sin,
            "cos": math.cos,
            "tan": math.tan,
            "exp": math.exp,
            "sqrt": math.sqrt,
            "log": math.log,
            "log10": math.log10,
            "pi": math.pi,
            "e": math.e,
            "abs": abs
        })

        # Integral
        resultado = Adaptive_Simpson(f, a, b, tolerancia)

        # Mostrar resultado
        print("\nResultado de la integral:", resultado)

    except (ValueError, TypeError, ZeroDivisionError,
            OverflowError, RecursionError) as error:
        print("Error:", error)

    finally:
        print("Operación finalizada.")

cuadratura_adaptable()
```
