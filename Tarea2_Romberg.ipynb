# 1. Introducción

La regla de Romberg es un método de integración numérica que permite
aproximar el valor de una integral definida.

El método utiliza aproximaciones obtenidas mediante la regla del
trapecio y posteriormente aplica la extrapolación de Richardson
para aumentar la precisión del resultado.

La integral definida que se desea aproximar se expresa como:

$$
\int_a^b f(x)\,dx
$$

El tamaño de paso utilizado en cada nivel se calcula mediante:

$$
h_i=\frac{b-a}{2^i}
$$

La primera aproximación se obtiene utilizando la regla del trapecio:

$$
R_{0,0}
=
\frac{b-a}{2}
\left[
f(a)+f(b)
\right]
$$

Para los siguientes niveles se incorporan nuevos puntos de
evaluación:

$$
R_{i,0}
=
\frac{1}{2}R_{i-1,0}
+
h_i
\sum_{\substack{k=1\\k\ impar}}^{2^i-1}
f(a+kh_i)
$$

Posteriormente se aplica la extrapolación de Richardson:

$$
R_{i,j}
=
R_{i,j-1}
+
\frac{
R_{i,j-1}-R_{i-1,j-1}
}{
4^j-1
}
$$

De esta manera se construye la tabla de Romberg y se obtiene una
aproximación cada vez más precisa de la integral.

La aproximación final se obtiene mediante:

$$
\boxed{
\int_a^b f(x)\,dx
\approx
R_{n-1,n-1}
}
$$

# 3. Implementación del algoritmo

A continuación, se presenta el código desarrollado en Python para
implementar el método de integración de Romberg descrito anteriormente.

El programa sigue los pasos del algoritmo y permite introducir la
función, los límites de integración y el número de niveles para
obtener la aproximación de la integral.

```python
from math import *

def romberg():

    try:

        # Entrada de datos
        funcion = input("Ingrese f(x) = ")
        a = float(input("Ingrese a: "))
        b = float(input("Ingrese b: "))
        n = int(input("Ingrese el número de niveles: "))

        # Validación de los datos
        if a >= b:
            raise ValueError("Debe cumplirse a < b")

        if n <= 0:
            raise ValueError("El número de niveles debe ser positivo")

        # Definición de la función
        f = lambda x: eval(funcion)

        # Creación de la tabla de Romberg
        R = [[0.0 for _ in range(n)] for _ in range(n)]

        # Cálculo del primer tamaño de paso
        h = b - a

        # Primera aproximación mediante la regla del trapecio
        R[0][0] = h * (f(a) + f(b)) / 2

        # Cálculo de los siguientes niveles
        for i in range(1, n):

            # Cálculo del tamaño de paso para cada nivel
            h = (b - a) / (2 ** i)

            # Inicialización de la suma
            suma = 0

            # Suma de los nuevos puntos de evaluación
            for k in range(1, 2 ** i, 2):
                suma += f(a + k * h)

            # Cálculo de la aproximación por trapecios
            R[i][0] = 0.5 * R[i - 1][0] + h * suma

            # Extrapolación de Richardson
            for j in range(1, i + 1):
                R[i][j] = R[i][j - 1] + \
                          (R[i][j - 1] - R[i - 1][j - 1]) / (4 ** j - 1)

        # Mostrar la tabla de Romberg
        print("\nTABLA DE ROMBERG\n")

        for i in range(n):
            for j in range(i + 1):
                print(round(R[i][j], 10), end="\t")
            print()

        # Mostrar la aproximación final de la integral
        print("\nIntegral =", round(R[n - 1][n - 1], 10))

    except Exception as e:
        print("Error:", e)

    finally:
        # Indica que el programa terminó su ejecución
        print("Operación finalizada")


# Llamada a la función
romberg()
```

La integración de Romberg se utiliza para aproximar

$$
\int_2^3 f(x)\,dx
$$

Si

$$
f(2)=0.51342
$$

$$
f(3)=0.36788
$$

$$
R_{3,1}=0.43687
$$

y

$$
R_{3,3}=0.43662
$$

encuentre

$$

```python
# Ejercicio 2 - Tarea 1
# Método de Romberg

# Datos del problema
f2 = 0.51342
f3 = 0.36788

R31 = 0.43687
R33 = 0.43662


# Cálculo de R(0,0)
R00 = (f2 + f3) / 2


# Cálculo de f(2.5)
f25 = (64 * R31 - 9 * R00 - 45 * R33) / 10


# Mostrar resultados
print("R(0,0) =", R00)
print("R(3,1) =", R31)
print("R(3,3) =", R33)

print("\nResultado:")
print("f(2.5) =", f25)

print("\nResultado aproximado:")
print("f(2.5) ≈", round(f25, 5))

print("\nOperación finalizada.")
```
